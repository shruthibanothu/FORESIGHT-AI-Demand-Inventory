```python
# Make copies so original data remains untouched
calendar_clean = cal_df.copy()
sales_clean = sales_df.copy()
sku_clean = sku_df.copy()
inventory_clean = invent_df.copy()


# Standardize date names
calendar_clean["date"] = pd.to_datetime(calendar_clean["date"])
sales_clean["Date"] = pd.to_datetime(sales_clean["Date"])
inventory_clean["Snapshot_Date"] = pd.to_datetime(
    inventory_clean["Snapshot_Date"])
sku_clean["Launch_Date"] = pd.to_datetime(sku_clean["Launch_Date"])
```

```python
# sorting the data
calendar_clean = calendar_clean.sort_values("date")

sales_clean = sales_clean.sort_values(
    ["SKU", "Date"]
)

inventory_clean = inventory_clean.sort_values(
    ["SKU", "Snapshot_Date"]
)

sku_clean = sku_clean.sort_values("SKU")
```

```python
# 4. Reset indexes
calendar_clean.reset_index(drop=True, inplace=True)
sales_clean.reset_index(drop=True, inplace=True)
inventory_clean.reset_index(drop=True, inplace=True)
sku_clean.reset_index(drop=True, inplace=True)
```

```python
master_df = (
    sales_clean
    .rename(columns={"Date": "date"})
    .merge(sku_clean, on="SKU", how="left")
    .merge(calendar_clean, on="date", how="left")
)

print(master_df.shape)
master_df.head()
```

```python
nums =[1,7,14,28]

for i in nums:
  master_df["lag_"+str(i)] = master_df.groupby("SKU")["Units_Sold"].shift(i)

```

```python
master_df = master_df.sort_values(["SKU", "date"])
# Rolling demand features (shift first to avoid data leakage)
grouped = master_df.groupby("SKU")["Units_Sold"]

master_df["rolling_mean_7"] = grouped.transform(
    lambda x: x.shift(1).rolling(7).mean()
)

master_df["rolling_mean_14"] = grouped.transform(
    lambda x: x.shift(1).rolling(14).mean()
)

master_df["rolling_mean_28"] = grouped.transform(
    lambda x: x.shift(1).rolling(28).mean()
)

master_df["rolling_std_7"] = grouped.transform(
    lambda x: x.shift(1).rolling(7).std()
)

# Remove rows where lag/rolling features aren't available
model_df = master_df.dropna(
    subset=[
        "lag_1", "lag_7", "lag_14", "lag_28",
        "rolling_mean_7", "rolling_mean_14",
        "rolling_mean_28", "rolling_std_7"
    ]
).copy()

print(model_df.shape)
```

```python
features = [
    "lag_1", "lag_7", "lag_14", "lag_28",
    "rolling_mean_7", "rolling_mean_14",
    "rolling_mean_28", "rolling_std_7",
    "Price", "Promotion",
    "month", "quarter", "week",
    "day_of_week", "is_weekend",
    "is_holiday", "is_promotion"
]

X = model_df[features]
y = model_df["Units_Sold"]

model_df = master_df.dropna(subset=features).copy()

# 2. Encode categorical features
# ============================================

model_df = pd.get_dummies(
    model_df,
    columns=["quarter", "day_of_week"],
    drop_first=True
)

features_encoded = [
    col for col in model_df.columns
    if col in features or
    col.startswith("quarter_") or
    col.startswith("day_of_week_")
]

```

```python
split_date = pd.Timestamp("2025-07-01")

train = model_df[model_df["date"] < split_date]
test = model_df[model_df["date"] >= split_date]

X_train = train[features_encoded]
y_train = train["Units_Sold"]

X_test = test[features_encoded]
y_test = test["Units_Sold"]

print("Train:", X_train.shape)
print("Test :", X_test.shape)
```

```python
from sklearn.ensemble import RandomForestRegressor

model = RandomForestRegressor(
    n_estimators=200,
    max_depth=20,
    random_state=42,
    n_jobs=-1
)
```

```python
model.fit(X_train, y_train)

y_pred = model.predict(X_test)


from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))

mask = y_test > 0
mape = np.mean(
    np.abs((y_test[mask] - y_pred[mask]) / y_test[mask])
) * 100

print(f"MAE  : {mae:.2f}")
print(f"RMSE : {rmse:.2f}")
print(f"MAPE : {mape:.2f}%")
```

```python
forecast_df = test[
    ["date", "SKU", "Product_Name", "Category"]
].copy()

forecast_df["Actual_Demand"] = y_test.values
forecast_df["Forecast_Demand"] = np.maximum(y_pred, 0)

print(forecast_df.head())
```

```python
importance = pd.DataFrame({
    "Feature": features_encoded,
    "Importance": model.feature_importances_
}).sort_values("Importance", ascending=False)

print("\nTop Features:")
print(importance.head(10))
```

```python
# ============================================
# INVENTORY INTELLIGENCE
# ============================================

# Use the latest inventory snapshot for each SKU
latest_inventory = (
    inventory_clean
    .sort_values("Snapshot_Date")
    .groupby("SKU")
    .tail(1)
    .copy()
)

# Average forecast demand per SKU
sku_forecast = (
    forecast_df
    .groupby("SKU")
    .agg(
        Forecast_Daily_Demand=("Forecast_Demand", "mean"),
        Actual_Daily_Demand=("Actual_Demand", "mean")
    )
    .reset_index()
)

# Merge inventory + forecast
inventory_intelligence = latest_inventory.merge(
    sku_forecast,
    on="SKU",
    how="left"
)

# --------------------------------------------
# Expected demand during lead time
# --------------------------------------------

inventory_intelligence["Lead_Time_Demand"] = (
    inventory_intelligence["Forecast_Daily_Demand"]
    * inventory_intelligence["Lead_Time_Days"]
)

# --------------------------------------------
# Stock coverage in days
# --------------------------------------------

inventory_intelligence["Stock_Coverage_Days"] = (
    (
        inventory_intelligence["Current_Stock"]
        + inventory_intelligence["On_Order"]
    )
    / inventory_intelligence["Forecast_Daily_Demand"].replace(0, np.nan)
)

# --------------------------------------------
# Inventory risk
# --------------------------------------------

def calculate_risk(row):
    available_stock = row["Current_Stock"] + row["On_Order"]
    lead_time_demand = row["Lead_Time_Demand"]
    reorder_point = row["Reorder_Point"]

    if available_stock <= reorder_point:
        return "HIGH"
    elif available_stock <= lead_time_demand:
        return "MEDIUM"
    else:
        return "LOW"

inventory_intelligence["Risk_Level"] = (
    inventory_intelligence.apply(calculate_risk, axis=1)
)

# --------------------------------------------
# Recommended reorder quantity
# --------------------------------------------

inventory_intelligence["Recommended_Order_Qty"] = np.maximum(
    inventory_intelligence["Reorder_Point"]
    - (
        inventory_intelligence["Current_Stock"]
        + inventory_intelligence["On_Order"]
    ),
    0
)

# --------------------------------------------
# Display results
# --------------------------------------------

inventory_intelligence = inventory_intelligence.merge(
    sku_clean[
        ["SKU", "Product_Name", "Category", "Selling_Price",
         "Cost_Price", "Gross_Margin_Per_Unit"]
    ],
    on="SKU",
    how="left"
)

print(inventory_intelligence[
    [
        "SKU",
        "Product_Name",
        "Current_Stock",
        "On_Order",
        "Forecast_Daily_Demand",
        "Lead_Time_Days",
        "Stock_Coverage_Days",
        "Risk_Level",
        "Recommended_Order_Qty"
    ]
].head(10))
```

```python
# ============================================
# FORESIGHT DECISION ENGINE
# ============================================

df = inventory_intelligence.copy()

# Available inventory
df["Available_Stock"] = df["Current_Stock"] + df["On_Order"]

# Estimated days until stockout
df["Days_Until_Stockout"] = (
    df["Available_Stock"] /
    df["Forecast_Daily_Demand"].replace(0, np.nan)
)

# Safety-stock adjusted target
df["Target_Stock"] = (
    df["Lead_Time_Demand"] + df["Safety_Stock"]
)

# Excess inventory
df["Excess_Stock"] = np.maximum(
    df["Available_Stock"] - df["Target_Stock"],
    0
)

# Final recommended order quantity
df["Recommended_Order_Qty"] = np.maximum(
    df["Target_Stock"] - df["Available_Stock"],
    0
)

# Business action
def get_action(row):
    if row["Risk_Level"] == "HIGH":
        return "REORDER NOW"
    elif row["Risk_Level"] == "MEDIUM":
        return "MONITOR"
    elif row["Excess_Stock"] > row["Forecast_Daily_Demand"] * 30:
        return "OVERSTOCK"
    else:
        return "NO ACTION"

df["Recommended_Action"] = df.apply(get_action, axis=1)

# Sort most urgent first
df = df.sort_values(
    ["Risk_Level", "Recommended_Order_Qty"],
    ascending=[True, False]
)

# Final FORESIGHT table
foresight_output = df[
    [
        "SKU",
        "Product_Name",
        "Category",
        "Current_Stock",
        "On_Order",
        "Forecast_Daily_Demand",
        "Lead_Time_Days",
        "Safety_Stock",
        "Reorder_Point",
        "Stock_Coverage_Days",
        "Days_Until_Stockout",
        "Excess_Stock",
        "Risk_Level",
        "Recommended_Order_Qty",
        "Recommended_Action"
    ]
].copy()

print(foresight_output.head(15))
```
